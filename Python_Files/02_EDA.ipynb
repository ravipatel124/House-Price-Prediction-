# Step 7 : Univariate Analysis of price column

## Histogram → Histogram shows the Distribution of Numerical Data

## 7.1 : Histogram of Price

```python
plt.figure(figsize=(10,5))

sns.histplot(df['price'], bins=35, kde=True)

plt.title("Distribution of House Prices")
plt.xlabel("Price (Lakhs)")
plt.ylabel("Number of Houses")
plt.show()
```

### Most houses are concentrated in the lower price range, while a few houses have very high prices. The distribution is right-skewed.

## 7.2 : Boxplot of Price

## Boxplot → Boxplot shows Outliers in Numerical Data

```python
plt.figure(figsize=(10,5))

sns.boxplot(x=df['price'])
plt.title("Boxplot of House Prices")
```

### Most house prices are within a normal range, while a few houses have very high prices. These high-priced houses appear as outliers.

## 7.3 : Analysis on BHK column

```python
plt.figure(figsize=(10,5))

sns.countplot(x='bhk', data=df)
plt.title("Distribution of BHK")
plt.xlabel("bhk")
plt.ylabel("Number of Houses")
```

### Most houses in the dataset have 2 BHK or 3 BHK. Houses with more BHK are less common.

## 7.4 :  Analysis On Bathroom

```python
plt.figure(figsize=(10,5))

sns.countplot(x='bath', data=df)
plt.title("Distribution of Bathrooms")
plt.xlabel("Number of Bathrooms")
plt.ylabel("Number of Houses")
```

### Most houses have 2 or 3 bathrooms. Houses with 4 or more bathrooms are much fewer.

## 7.5 : Analysis On Balcony

```python
plt.figure(figsize=(10,5))
sns.countplot(x='balcony', data=df)

plt.title("Distribution of Balconies")
plt.xlabel("Number of Balconies")
plt.ylabel("Number of Houses")
```

### Most houses have 1 or 2 balconies, while houses with more balconies are less common.

## 7.6 : Univariate Analysis - Area Type

```python
plt.figure(figsize=(10,5))
sns.countplot(x='area_type', data=df)

plt.title("Distribution of Area Types")
plt.xlabel("Number of Houses")
plt.ylabel("Area Type")
```

### Most properties belong to the Super built-up Area category, while other area types have fewer properties.

## 7.5: Top 10 Locations

```python
top_locations = df['location'].value_counts().head(10)

plt.figure(figsize=(10,6))
sns.barplot(x=top_locations.values,y=top_locations.index)

plt.title("Top 10 Locations")
plt.xlabel("Number of Houses")
plt.ylabel("Location")
```

### Whitefield has the highest number of houses in the dataset. Sarjapur Road and Electronic City also have many houses, while the other locations have fewer houses.

# Step 8 : Bivariate Analysis

## 8.1 : Price vs Total Square Feet

```python
plt.figure(figsize=(10,6))

sns.scatterplot(x='total_sqft',y='price',data=df,alpha=0.6,s=100)
plt.title("Price vs Total Square Feet")
plt.xlabel("Total Square Feet")
plt.ylabel("Price (Lakhs)")
```

### As the total square feet increases, the house price also generally increases. However, a few houses have unusually high or low prices, which may be outliers.

## 8.2  : Price vs BHK

```python
plt.figure(figsize=(10,6))

sns.boxplot(x='bhk',y='price',data=df)
plt.title("Price vs BHK")
plt.xlabel("BHK")
plt.ylabel("Price (Lakhs)")
```

### Houses with more bedrooms generally have higher prices. However, some variations exist due to location and other property features.

## 8.3 : Price vs Bathrooms

```python
plt.figure(figsize=(10,6))

sns.boxplot(x='bath',y='price',data=df)
plt.title("Price vs Bathrooms")
plt.xlabel("Bathrooms")
plt.ylabel("Price (Lakhs)")
```

### Houses with more bathrooms generally have higher prices, although a few outliers are present.

## 8.4 : Price vs Area Type

```python
plt.figure(figsize=(10,6))
sns.boxplot(x='area_type',y='price',data=df)

plt.title("Price vs Area Type")
plt.xlabel("Area Type")
plt.ylabel("Price (Lakhs)")
plt.xticks(rotation=0)
```

### The box plot shows that house prices vary across different area types. Some houses have much higher prices than the others, indicating outliers.

# Step 9 : Correlation Analysis

### Correlation measures the relationship between numerical features.

## 9.1 : Correlation Heatmap

```python
correlation = df.select_dtypes(include=['number']).corr()
correlation
```

### This table shows how strongly numerical features are related to each other.

```python
plt.figure(figsize=(8,6))
sns.heatmap(correlation,annot=True,cmap='coolwarm',fmt=".4f")
plt.title("Correlation Heatmap")

plt.show()
```
